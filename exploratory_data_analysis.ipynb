# Exploratory Data Analysis of UN data
We have UN open source data. This notebook will have an inital look into the data sources we have and visualise the information we have.

```python
# import packages
import pandas as pd
import matplotlib.pyplot as plt 

%matplotlib inline
```

```python
# data paths
folder_path = "C:/Users/heath/Documents/self_projeects/repositories/un_data_analysis/input/UN"
file_name = "SYB68_230_202511_GDP and GDP Per Capita.csv"
filepath = f"{folder_path}/{file_name}"
```

```python
# initialise data frame
df = pd.read_csv(filepath, encoding="latin-1", skiprows=1)
cols_to_change = {
    "Region/Country/Area": "region_code",
    "Unnamed: 1": "region_name",
}
df = df.rename(cols_to_change, axis=1)
df.columns = [c.lower() for c in df.columns]
df.head(2)
```

```python
def visualise_gdp_per_capita(df: pd.DataFrame, country: str):
    """Collect and visualise per capita GDP of a given country."""
    df_country = df.loc[df["region_name"]==country]
    df_country_per_capita = df_country.loc[df_country["series"]=="GDP per capita (US dollars)"]
    # visual
    plt.plot(df_country_per_capita["year"], df_country_per_capita["value"])
    # plt.title(f"GDP per capita of {country}")
    # plt.show()

visualise_gdp_per_capita(df=df, country="Total, all countries or areas")
```
